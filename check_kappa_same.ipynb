```python
import numpy as np

new = np.load("kappa_samples_stratified_lowpH.npz")
old = np.load("kappa_samples_stratified_lowpH_old.npz")

print("New keys:", new.files)
print("Old keys:", old.files)
```

```python
print("New first 5:", new["kappa_samples"][:5])
print("Old first 5:", old["kappa_samples"][:5])
```

```python
print("New shape:", new["kappa_samples"].shape, new["kappa_samples"].dtype)
print("Old shape:", old["kappa_samples"].shape, old["kappa_samples"].dtype)
```

```python
np.array_equal(new["kappa_samples"], old["kappa_samples"])
```

```python
print("Means:", new["kappa_samples"].mean(), old["kappa_samples"].mean())
print("Stds:", new["kappa_samples"].std(), old["kappa_samples"].std())
print("Mins:", new["kappa_samples"].min(), old["kappa_samples"].min())
print("Maxs:", new["kappa_samples"].max(), old["kappa_samples"].max())
```

```python
import matplotlib.pyplot as plt

plt.hist(np.log10(new["kappa_samples"]), bins=100, alpha=0.5, label="new", density=True)
plt.hist(np.log10(old["kappa_samples"]), bins=100, alpha=0.5, label="old", density=True)
plt.legend()
plt.title("Low pH kappa samples comparison")
plt.show()
```

```python
diff_idx = np.where(new["kappa_samples"] != old["kappa_samples"])[0]
print("Number of differing entries:", len(diff_idx))
```
